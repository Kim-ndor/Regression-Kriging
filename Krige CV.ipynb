```python
# Importing libraries
# conda install conda-forge::scikit-learn
# conda install pandas
# conda install conda-forge::geopandas
import numpy as np
import pandas as pd
from sklearn.model_selection import GridSearchCV, train_test_split
from pykrige.rk import Krige
# Scaling the input features
from sklearn.preprocessing import StandardScaler
# To ignore warning messages
import warnings
warnings.filterwarnings("ignore")
```

```python
# 2. Loading the dataset
data = pd.read_csv('E:/VData/SData/Wells3.csv')
```

```python
data.head()
```

```python
# 3. Data preparation
```

```python
# The predictor variables
X = []
X.append(data['T'])
X.append(data['Depth'])
X = np.array(X).T
```

```python
X.shape
```

```python
# The target
```

```python
y = data['NO3']
```

```python
################# 2D Kriging optimization, our dataset is constituted of 2D points #####################
```

```python
# 1. # a. Defining the 2d parameters dictionary   

param_dict = {
    'method': ['ordinary', 'universal'],
    'variogram_model': ['linear', 'power', 'gaussian', 'spherical', 'exponential'],
}
```

```python
# 2. Running the GridSearchCV
```

```python
# a. Creating an estimator
estimator = GridSearchCV(Krige(), param_dict, verbose = True, return_train_score = True)
```

```python
# b. Fitting the estimator
estimator.fit(X, y)
```

```python
# c. Outpuuting the results
if hasattr(estimator, "best_score_"):
    print("best_params = ", estimator.best_params_)
    print("best_score R² = {:.2f}".format(estimator.best_score_))
print("\nCV results::")
if hasattr(estimator, "cv_results_"):
    for key in [
        "mean_test_score",
        "mean_train_score",
        "param_method",
        "param_variogram_model",
    ]:
        print(" - {} : {}".format(key, estimator.cv_results_[key]))
```
